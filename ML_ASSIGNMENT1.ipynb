**Word Frequency Vector (Bag-of-Words)** :

**Predection** : The word that will have a count of 2 is:

All the other words appear only once, so their count will be 1.

```python
def bag_of_words_vector(tokens, vocabulary):              # Create a function to count each word
    vector = np.array([np.sum(np.array(tokens) == word)  # Count how many times each word appears
                       for word in vocabulary])
    return vector                                         # Return the word-count vector
# Build vocabulary for tokens2
tokens2_array = np.array(tokens2)                         # Convert tokens2 into a NumPy array
vocabulary2 = np.unique(tokens2_array)                   # Get unique words
vec2 = bag_of_words_vector(tokens2, vocabulary2)         # Create the bag-of-words vector
print("Vocabulary:", vocabulary2)                        # Display the vocabulary
print("Vector:", vec2)                                   # Display the word counts
```

**Explanation** : I am creating a Bag of Words vector. It checks each word in the vocabulary and counts how many times that word appears in the sentence. Since every word in tokens2 appears only once, the vector contains 1 for each word

**Comparing Two Sentences (Cosine Similarity)** :

```python
# Similar sentences
sentence1 = "I like playing football"                    # First sentence
sentence2 = "I enjoy playing football"                  # Similar sentence
# Different sentences
sentence3 = "I like playing football"                    # Third sentence
sentence4 = "The weather is very cold today"            # Completely different sentence
# Create a function to calculate cosine similarity
def cosine_similarity(sentence_a, sentence_b):

    tokens_a = sentence_a.lower().split()               # Convert first sentence into lowercase words
    tokens_b = sentence_b.lower().split()               # Convert second sentence into lowercase words

    vocabulary = np.unique(np.array(tokens_a + tokens_b))  # Create a list of unique words from both sentences

    vec_a = bag_of_words_vector(tokens_a, vocabulary)   # Convert first sentence into a word-count vector
    vec_b = bag_of_words_vector(tokens_b, vocabulary)   # Convert second sentence into a word-count vector

    dot_product = np.dot(vec_a, vec_b)                  # Calculate the dot product of the two vectors

    norm_a = np.linalg.norm(vec_a)                      # Calculate the length of the first vector
    norm_b = np.linalg.norm(vec_b)                      # Calculate the length of the second vector

    similarity = dot_product / (norm_a * norm_b)        # Calculate the cosine similarity

    return similarity                                   # Return the similarity score
# Calculate similarity between the similar sentences
similar_score = cosine_similarity(sentence1, sentence2)  # Get similarity score
# Calculate similarity between the different sentences
different_score = cosine_similarity(sentence3, sentence4) # Get similarity score
# Display the results
print("Similarity score:", similar_score)               # Print score for similar sentences
print("Different score:", different_score)             # Print score for different sentences
```

**explanation** :
The first pair talks about playing football, so they should have a higher similarity score.
The second pair talks about football and weather, so they should have a lower similarity score.

**Reflection** : Cosine similarity tells us how similar two sentences are.
If two sentences have similar words, their value is close to 1.
If two sentences are very different, their value is close to 0.
A cosine similarity of exactly 1 means the two vectors are completely similar and point in the same direction.

```python
### #Mini Project: Simple Text Similarity Checker (NLP Track)
```

```python
import numpy as np                                      # Import NumPy


# Create three sentences
sentence1 = "I love playing football"                  # First sentence
sentence2 = "I enjoy playing football"                 # Similar to sentence 1
sentence3 = "The weather is very cold today"           # Different sentence


# Convert each sentence into separate words
tokens1 = sentence1.lower().split()                    # Tokenize sentence 1
tokens2 = sentence2.lower().split()                    # Tokenize sentence 2
tokens3 = sentence3.lower().split()                    # Tokenize sentence 3


# Create one common vocabulary from all three sentences
all_tokens = tokens1 + tokens2 + tokens3               # Combine all words
vocabulary = np.unique(np.array(all_tokens))           # Find unique words


# Create a function to convert words into word-count vectors
def bag_of_words_vector(tokens, vocabulary):           # Define the function
    vector = np.array([np.sum(np.array(tokens) == word) # Count each word
                       for word in vocabulary])
    return vector                                      # Return the vector


# Convert all three sentences into vectors
vec1 = bag_of_words_vector(tokens1, vocabulary)        # Vector for sentence 1
vec2 = bag_of_words_vector(tokens2, vocabulary)        # Vector for sentence 2
vec3 = bag_of_words_vector(tokens3, vocabulary)        # Vector for sentence 3


# Create a function to calculate cosine similarity
def cosine_similarity(vec_a, vec_b):                   # Define similarity function
    dot_product = np.dot(vec_a, vec_b)                 # Calculate dot product
    norm_a = np.linalg.norm(vec_a)                     # Find length of first vector
    norm_b = np.linalg.norm(vec_b)                     # Find length of second vector

    return dot_product / (norm_a * norm_b)              # Calculate cosine similarity


# Calculate similarity between sentence 1 and sentence 2
similarity_12 = cosine_similarity(vec1, vec2)          # Compare sentence 1 and 2

# Calculate similarity between sentence 1 and sentence 3
similarity_13 = cosine_similarity(vec1, vec3)          # Compare sentence 1 and 3

# Calculate similarity between sentence 2 and sentence 3
similarity_23 = cosine_similarity(vec2, vec3)           # Compare sentence 2 and 3


# Display the similarity scores
print("Sentence 1 and 2:", similarity_12)              # Print score for pair 1-2
print("Sentence 1 and 3:", similarity_13)              # Print score for pair 1-3
print("Sentence 2 and 3:", similarity_23)              # Print score for pair 2-3


# Store all similarity scores
scores = [similarity_12, similarity_13, similarity_23]  # Put scores into a list

# Find the highest similarity score
highest_score = max(scores)                            # Find the largest score


# Find which pair has the highest score
if highest_score == similarity_12:                      # Check if pair 1-2 is highest
    print("Most similar: Sentence 1 and Sentence 2")

elif highest_score == similarity_13:                   # Check if pair 1-3 is highest
    print("Most similar: Sentence 1 and Sentence 3")

else:                                                  # Otherwise pair 2-3 is highest
    print("Most similar: Sentence 2 and Sentence 3")
```

```python
**Explanation** : I take three sentences. Then I split each sentence into individual words. I create one common vocabulary containing all the unique words from the three sentences. After that, I convert each sentence into a Bag-of-Words vector, where the numbers represent how many times each word appears. Then I calculate cosine similarity between every pair of sentences. Finally, I compare the three similarity scores and print the pair with the highest score as the most similar sentences.
```
